## 1) What is Matplotlib, really?

**Matplotlib** is a Python library for creating **2D plots** (and some 3D via addons).It's built around a concept called a **plotting state** (when using `pyplot`) and a more  “engineering-style” system called the **object-oriented (OO) API**.



### Two ways to use Matplotlib

#### 1. Pyplot style (quick & easy)
- You write commands like `plt.plot()`, `plt.title()`, etc.
- It behaves like a **“current plot” system**.

#### 2. Object-oriented style (professional & scalable)
- You explicitly create objects:
  - `fig, ax = plt.subplots()`
  - then use `ax.plot(...)`



> **Day 1** focuses on pyplot style because it’s easiest to start with.


<br>
<hr>

## 2) Why do we write: import matplotlib.pyplot as plt?

### What `pyplot` is

`matplotlib.pyplot` is a module that provides **MATLAB-like plotting functions**

<p>When you do:</p>

<pre><code>
import matplotlib.pyplot as plt
</code></pre>

you’re importing that module and renaming it to `plt` for convenience (standard convention).

**What pyplot does behind the scenes**

When you call pyplot functions like `plt.plot()`, Matplotlib:

* creates (or reuses) a **Figure** (the whole window/image)
* creates (or reuses) an **Axes** (the actual chart area with x/y)
* draws your plot onto those axes

So even if you never mention “Figure” or “Axes”, they exist.

## 3) The most important concept: Figure vs Axes

### ✅ Figure

- Think of a Figure like a blank canvas / page.
- A single figure can contain multiple plots (multiple axes/subplots).


### ✅ Axes

- An Axes is the actual plot area with:

    - x-axis + y-axis
    - grid
    - title
    - labels
    - the plotted line(s)

`Common confusion:`

Axes ≠ Axis

- Axis = x-axis or y-axis line + ticks
- Axes = the whole plotting region

## 5) What is a “Line2D” and what gets created?

When you call `plt.plot()`, Matplotlib creates a Line2D object internally.


<p>That line object stores:</p>


* data (x and y)
* line color
* style (solid/dashed/dotted)
* thickness
* markers (if any)
* label (for legend)

<p>Example:</p>

<div style="
  background-color:#f6f8fa;
  border-radius:6px;
  padding:12px;
  font-family:monospace;
  font-size:14px;
  width:fit-content;
">
  line = plt.plot(x, y)
</div>


<p>Actually returns a list of line objects (because plot can draw multiple lines).</p>

## 6) Why do we need plt.show()?

**In scripts**

If you run a `.py` file, Python finishes quickly. Without `plt.show()`:

* the window may never appear
* or may flash and close

So `plt.show()` tells Python:

> “Open the plot window and start the GUI event loop so the plot stays visible.”


**In Jupyter notebooks**

Sometimes plots auto-display, but `plt.show()` is still good practice (especially for clean control).

# 8) Title and Labels: what are they technically?

These are **text objects** attached to the axes:

* `plt.title(...)` sets the Axes title
* `plt.xlabel(...)` sets x-axis label
* `plt.ylabel(...)` sets y-axis label


<p>These help viewers interpret:</p>

* what x means
* what y means
* what the whole plot represents

# 9) Practicals for Day 1

### ✅ Practical 1: Plot 1–10 vs squares

```python
import matplotlib.pyplot as plt

x = list(range(1, 11))
y = [i**2 for i in x]

plt.plot(x, y)
plt.show()
```

## ✅ Practical 2: Change line color & style

```python
plt.plot(x, y, color='purple', linestyle='--', linewidth=2)
plt.show()
```

## ✅ Practical 3: Add title and axis labels

```python
plt.plot(x, y, color='purple', linestyle='--', linewidth=2)

plt.title("Squares from 1 to 10")
plt.xlabel("Number (x)")
plt.ylabel("Square (x²)")

plt.show()
```
