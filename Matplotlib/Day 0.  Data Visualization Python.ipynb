 <h1 align="center">1️⃣ What is Matplotlib? </h1>

<br>

<p> <b> Matplotlib </b> is a <b>Python library </b> used to <b> create graphs and charts.</b></p>

<p>It helps us <b> visualize data </b> using:</p>

* Line graphs
* Bar charts
* Pie charts
* Histograms
* Scatter plots

#### 📌 In simple words:

### 👉 Matplotlib = drawing graphs using Python

 <h1 align="center"> 2️⃣ Why do we learn Matplotlib? </h1>

<br>

<p>We learn Matplotlib because:</p>

- ✅ Numbers alone are <b> hard to understand</b>  
- ✅ Graphs make data <b> easy and clear </b> 
- ✅ Used in <b> Data Science, AI, ML, and Analytics </b> 
- ✅ Helps find <b> patterns, trends, and comparisons </b> 

<br>

* Table → confusing 😕

 <h1 align="center"> 4️⃣ What is Data Visualization? </h1>

<br>

 ➤ **Data visualization is the graphical representation of information and data.**

➤ **Visual elements like charts, graphs, and maps provide an accessible way to see and understand trends, outliers, and patterns in data.**

<br>

<P>📌 Simple definition:</p>
<p>👉 <b>Data Visualization = showing data using pictures</b></p>
<br>
<hr>

<ul style="list-style: none; font-size: 20px; line-height: 1.8;">

 <h1 align="center"> 5️⃣ Types of Data Visualization </h1>

<br>
  

<P> Here are the <b> most important types</b> 👇</P>

<br>

### 🔹 1. Line Chart

<p> 📈 Used for <b> trend over time</b></p>

* Stock price
* Temperature changes

<br>

### 🔹 2. Bar Chart

<p>📊 Used for <b>comparison</b></p>

* Student marks
* Sales of products

<br>

### 🔹 3. Pie Chart

<p>🥧 Used for <b> percentage or share</b></p>

* Market share
* Budget distribution

<br>

### 🔹 4. Scatter Plot

<p>🔵 Used for <b> relationship between two variables</b></p>

* Height vs Weight


<br>


### 🔹 5. Histogram

<p>🧱 Used for <b> data distribution</b></p>

* Exam score distribution
